```python
from enum import StrEnum, Enum


class ParameterType(StrEnum):
    INT = 'int'
    BOOL = 'bool'
    STRING = 'string'

class Parameter():
    def __init__(self,
                 id: int,
                 name: str,
                 url_name: str,
                 required: bool,
                 type: ParameterType,
                 meta,
                 multi: bool,
                 ui_component,
                 actions,
                 grouping,
                 labels,
                 values,
                 external_values_url,
                 range,
                 min_taxonomy_version: int,
                 hint,
                 image_url,
                 is_type: bool) -> None:
        self.id = id  # ID
        self.name = name  # Full name
        self.url_name = url_name  # Short name for url
        self.required = required  # Is required (True, False)
        self.type = type  # Type of param (int, bool, string)
        self.meta = meta  # ? ('set_free', 'set', 'free')
        self.multi = multi  # ? (True, False)
        self.ui_component = ui_component  # "{'fallback_type': None, 'type': None}", "{'fallback_type': 'multiselect_and', 'type': 'multiselect_and'}", "{'fallback_type': 'select_range', 'type': 'select_range'}", "{'fallback_type': 'multiselect_or', 'type': 'multiselect_or'}", "{'fallback_type': 'radio_button', 'type': 'radio_button'}", "{'fallback_type': 'multiselect_or', 'type': 'multiselect_chips_or'}", "{'fallback_type': 'radio_button', 'type': 'select_chips'}", "{'fallback_type': 'select', 'type': 'select_chips'}", "{'fallback_type': 'input_range', 'type': 'input_range'}", 'None', "{'fallback_type': 'select', 'type': 'select_search'}", "{'fallback_type': 'multiselect_chips_or', 'type': 'multiselect_chips_or'}", "{'fallback_type': 'select', 'type': 'select'}", "{'fallback_type': 'checkbox', 'type': 'new_checkbox'}", "{'fallback_type': 'checkbox', 'type': 'checkbox'}", "{'fallback_type': 'input', 'type': 'input'}"
        self.actions = actions  # ??
        self.grouping = grouping  # ??
        self.labels = labels  # Label(name) of paramenter and placeholder
        self.values = values  # Values with lables
        self.external_values_url = external_values_url  # ?? (Null)
        # Range of value (Null, ) {'default_lower': 3, 'default_upper': 7, 'lower': 1, 'step': 1, 'upper': 15}
        self.range = range
        # Version of taxonomy (0)
        self.min_taxonomy_version = min_taxonomy_version
        self.hint = hint  # Hint/подсказка
        self.image_url = image_url  # ?? (Null)
        self.is_type = is_type  # Is type (True, False)
```

```python
import sys
sys.path.append('../')
from categories import *


def parse_categories(data):
    """Parse the categories data into objects"""
    categories = []
    category_list = data.get('categories', [])

    for categ in category_list:
        subcategories = []
        for subcateg in categ['subcategories']:
            subcategories.append(
                Subcategorie(
                    int(subcateg['id']),
                    subcateg['name'],
                    subcateg['name_ru'],
                    subcateg['name_by'],
                    int(subcateg['order']),
                    int(subcateg['parent']),
                )
            )

        categories.append(
            Categorie(
                int(categ['id']),
                categ['name'],
                categ['name_ru'],
                categ['name_by'],
                int(categ['version']),
                int(categ['order']),
                subcategories
            )
        )

    return categories
```

```python
params = []
existing_ids = set()

for categ in categs:
    response = requests.get(f'https://api.kufar.by/taxonomy-proxy/v1/dispatch?routing=web_generalist&application=ad_listing&platform=web&taxonomy-version=1&parent={categ.id}')
    data = response.json()
    params_json = data['metadata']['parameters']['refs']
    for param_id, obj in params_json.items():
        if param_id not in existing_ids:
            parameter = Parameter(
                id=obj['variation_id'],
                name=obj['name'],
                url_name=obj['url_name'],
                required=obj['required'],
                type=obj['type'],
                meta=obj['meta'],
                multi=obj['multi'],
                ui_component=obj['ui_component'],
                actions=obj['actions'],
                grouping=obj['grouping'],
                labels=obj['labels'],
                values=obj['values'],
                external_values_url=obj['external_values_url'],
                range=obj['range'],
                min_taxonomy_version=obj['min_taxonomy_version'],
                hint=obj['hint'],
                image_url=obj['image_url'],
                is_type=obj['is_type']
            )
            params.append(parameter)
            existing_ids.add(param_id)
    # params_id.extend(ids)
```

```python
# Initializing sets for unique values
unique_names = set()
unique_url_names = set()
unique_required = set()
unique_types = set()
unique_meta = set()
unique_multi = set()
unique_ui_components = set()
unique_actions = set()
unique_groupings = set()
unique_labels = set()
unique_values = set()
unique_external_values_urls = set()
unique_ranges = set()
unique_min_taxonomy_versions = set()
unique_hints = set()
unique_image_urls = set()
unique_is_types = set()

for parameter in params:
    unique_names.add(parameter.name)
    unique_url_names.add(parameter.url_name)
    unique_required.add(parameter.required)
    unique_types.add(parameter.type)
    unique_meta.add(parameter.meta)
    unique_multi.add(parameter.multi)
    unique_ui_components.add(str(parameter.ui_component))
    if parameter.actions:
        unique_actions.update(str(action.items()) for action in parameter.actions)
    unique_groupings.add(str(parameter.grouping))
    unique_labels.add(str(parameter.labels.items()))
    if parameter.values:
        unique_values.update(str(value.items()) for value in parameter.values)
    unique_external_values_urls.add(parameter.external_values_url)
    unique_ranges.add(str(parameter.range))
    unique_min_taxonomy_versions.add(parameter.min_taxonomy_version)
    unique_hints.add(str(parameter.hint))
    unique_image_urls.add(parameter.image_url)
    unique_is_types.add(parameter.is_type)

# Converting sets to lists if needed
unique_names = list(unique_names)
unique_url_names = list(unique_url_names)
unique_required = list(unique_required)
unique_types = list(unique_types)
unique_meta = list(unique_meta)
unique_multi = list(unique_multi)
unique_ui_components = list(unique_ui_components)
unique_actions = list(unique_actions)
unique_groupings = list(unique_groupings)
unique_labels = list(unique_labels)
unique_values = list(unique_values)
unique_external_values_urls = list(unique_external_values_urls)
unique_ranges = list(unique_ranges)
unique_min_taxonomy_versions = list(unique_min_taxonomy_versions)
unique_hints = list(unique_hints)
unique_image_urls = list(unique_image_urls)
unique_is_types = list(unique_is_types)
```
